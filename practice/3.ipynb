```python
from pathlib import Path
```

```python
path1 = Path("./../dl/3/fashion-mnist_train.csv")
path2 = Path("./../dl/3/fashion-mnist_test.csv")
```

```python
import pandas as pd
```

```python
train_df = pd.read_csv(path1)
train_df
```

```python
train_df.shape
```

```python
test_df = pd.read_csv(path2)
test_df
```

```python
test_df.shape
```

```python
TRAIN_DF_SIZE = 6000
TEST_DF_SIZE = 1000
```

```python
train_df = train_df.sample(TRAIN_DF_SIZE, random_state=42)
train_df.reset_index(drop=True)
train_df
```

```python
test_df = test_df.sample(TEST_DF_SIZE, random_state=42)
test_df = test_df.reset_index(drop=True)
test_df
```

```python
x_train = train_df.drop("label", axis = 1)
```

```python
x_train = x_train.values.reshape(-1, 28, 28,1)
```

```python
x_train = x_train.astype("float64") / 255.0
x_train
```

```python
y_train = train_df.label
y_train = y_train.values.astype("int64")
y_train
```

```python
x_test = test_df.drop("label", axis = 1)
```

```python
x_test = x_test.values.reshape(-1, 28, 28, 1)
```

```python
x_test = x_test.astype("float64") / 255.0
```

```python
x_test
```

```python
y_test = test_df.label
y_test = y_test.values.astype("int32")
y_test
```

```python
import tensorflow as tf
```

```python
classNames = [
    "t shirt",
    "trouser",
    "pullover",
    "dress",
    "coat",
    "sandle",
    "suit",
    "sneaker",
    "bag",
    "ankle"
]
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 4))

for i in range(0, 10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_train[i], cmap="gray")
    plt.title(classNames[y_train[i]])
    plt.axis("off")

plt.show()
```

```python
from tensorflow.keras import layers, Sequential
```

```python
model = Sequential([
    layers.Input(shape=(28, 28, 1)),
    layers.Conv2D(32, kernel_size=(3, 3), activation="relu"),
    layers.MaxPool2D(pool_size=(2, 2)),
    layers.Conv2D(64, kernel_size=(3, 3), activation="relu"),
    layers.MaxPool2D(pool_size=(2, 2)),
    layers.Flatten(),
    layers.Dense(64, activation="relu"),
    layers.Dropout(0.3),
    layers.Dense(10, activation='softmax')
])
                     
    
```

```python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
model.summary()
```

```python
history = model.fit(
    x_train,
    y_train,
    epochs=5,
    validation_split=0.2,
    batch_size=128,
)
```

```python
plt.figure(figsize=(8, 4))
plt.plot(history.history["accuracy"], label="Training accuracy")
plt.plot(history.history["val_accuracy"], label="Validation accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Fashion-MNIST CNN Accuracy")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
from sklearn.metrics import classification_report
```

```python
test_loss, test_accuracy = model.evaluate(x_test, y_test)
print(f"Test accuracy: {test_accuracy:.4f}")

y_prob = model.predict(x_test, verbose=0)
y_pred = np.argmax(y_prob, axis=1)

print(classification_report(
    y_test,
    y_pred,
    target_names=classNames
))
```
