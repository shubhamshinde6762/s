```python
from pathlib import Path
```

```python
train_path = Path("./../dl/3/fashion-mnist_train.csv")
test_path = Path("./../dl/3/fashion-mnist_test.csv")
```

```python
import pandas as pd
```

```python
train_df = pd.read_csv(train_path)
train_df
```

```python
test_df = pd.read_csv(test_path)
test_df
```

```python
TRAIN_DF_SIZE = 6000
TEST_DF_SIZE = 1000
```

```python
train_df = train_df.sample(TRAIN_DF_SIZE)
```

```python
train_df = train_df.reset_index(drop=True)
train_df
```

```python
test_df = test_df.sample(TRAIN_DF_SIZE)
test_df = test_df.reset_index(drop=True)
test_df
```

```python
x_train = train_df.drop("label", axis=1)
```

```python
x_train = x_train.values
```

```python
x_train = x_train.reshape(-1, 28, 28, 1)
```

```python
x_train = x_train.astype('float64') / 255.0
```

```python
y_train = train_df.label.values.astype('int64')
y_train
```

```python
x_train
```

```python
x_test = test_df.drop('label', axis=1)
```

```python
x_test = x_test.values
```

```python
x_test = x_test.reshape(-1, 28, 28, 1)
```

```python
x_test = x_test.astype('float64') / 255.0
```

```python
y_test = test_df.label.values.astype('int64')
y_test
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8, 4))

for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_train[i], cmap="gray")
    plt.title(str(y_train[i]))

plt.show()
```

```python
class_names = [
    't-shirt',
    'trouser',
    'pullover',
    'dress',
    'coat',
    'sandle',
    'shirt',
    'sneaker',
    'bag',
    'ankle'
]
```

```python
plt.figure(figsize=(8, 4))

for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_train[i], cmap="gray")
    plt.title(class_names[y_train[i]])

plt.show()
```

```python
import tensorflow as tf
from tensorflow.keras import Sequential, layers
```

```python
model = Sequential([
    layers.Input(shape=(28, 28, 1)),
    layers.Conv2D(32, kernel_size=(3, 3), activation='relu'),
    layers.MaxPooling2D(pool_size=(2, 2)),
    layers.Conv2D(64, kernel_size=(3, 3), activation='relu'),
    layers.MaxPooling2D(pool_size=(2, 2)),
    layers.Flatten(),
    layers.Dense(64, activation="relu"),
    layers.Dropout(0.3),
    layers.Dense(10, activation='softmax')
])
```

```python
model.compile(
    optimizer='adam',
    metrics=['accuracy'],
    loss='sparse_categorical_crossentropy'
)
```

```python
history = model.fit(
    x_train,
    y_train,
    epochs=7,
    batch_size=128,
    validation_split=0.2
)
```

```python
y_pred = model.predict(x_test)
y_pred
```

```python
import numpy as np

y_pred = np.argmax(y_pred, axis=1)
y_pred
```

```python
y_test
```

```python
plt.figure(figsize=(8, 8))

for i in range(20):    
    plt.subplot(4, 5, i + 1)
    plt.imshow(x_train[i], cmap="gray")
    color = 'green' if y_test[i] == y_pred[i] else 'red'
    plt.title(f"A : {class_names[y_test[i]]}\n P : {class_names[y_pred[i]]}", color=color)

plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred, target_names=class_names))
```

```python
plt.figure(figsize=(8, 4))
plt.plot(history.history["accuracy"], label="Training accuracy")
plt.plot(history.history["val_accuracy"], label="Validation accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Fashion-MNIST CNN Accuracy")
plt.legend()
plt.grid(True)
plt.show()
```
